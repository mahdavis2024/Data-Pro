<div style="background-color: lime;color:black; padding: 2px">
<p>First, extract the raw text after the style ID and inspect its values. This lets us identify unusual size labels before deciding whether they should be standardized. Products whose product_id equals style_id have no size suffix.</p>
</div>

```python
def extract_raw_size_suffix(row):
    product_id = row["product_id"]
    style_id = row["style_id"]

    if pd.isna(product_id) or pd.isna(style_id):
        return pd.NA

    product_id = str(product_id).strip()
    style_id = str(style_id).strip()

    for separator in ["_", "-"]:
        prefix = style_id + separator

        if product_id.casefold().startswith(prefix.casefold()): # case-insensitive comparison
            return product_id[len(prefix):].strip().upper()     # extract the suffix after the style_id and separator

    return pd.NA

edf["raw_size_suffix"] = edf.apply(extract_raw_size_suffix, axis=1).astype("string")
```

```python
# Inspect suffix values before standardizing any labels
edf["raw_size_suffix"].value_counts(dropna=False).head(20)
```

```python
# Products with no extracted size suffix; these have product_id equal to style_id
no_size_suffix = edf["raw_size_suffix"].isna()
edf.loc[no_size_suffix, ["product_id", "style_id", "raw_size_suffix"]]
```

<div style="background-color: lime;color:black; padding: 2px">
<p>The raw suffix counts reveal two unusual labels: <b>MM</b> and <b>2XLL</b>. These appear to be typos for <b>M</b> and <b>2XL</b>, so I standardize them before summarizing the size distribution. The raw suffix column is removed after this step.</p>
</div>


```python
# Standardize the two unusual labels identified in the raw suffix counts
size_order = ["XS", "S", "M", "L", "XL", "2XL", "3XL"]
size_aliases = {"MM": "M", "2XLL": "2XL"}

edf["size"] = edf["raw_size_suffix"].replace(size_aliases)
edf["size"] = edf["size"].where(edf["size"].isin(size_order), pd.NA)

# Define the order of sizes for sorting and plotting
edf["size"] = pd.Categorical(edf["size"], categories=size_order, ordered=True)

edf = edf.drop(columns="raw_size_suffix")
edf.head(3)
```

```python
size_order_with_missing = size_order + ["No size"]
size_counts = (
    edf["size"]
    .astype("string")
    .fillna("No size")
    .value_counts()
    .reindex(size_order_with_missing, fill_value=0)
)
size_summary = pd.DataFrame({
    "product_id_count": size_counts,
    "percentage": (size_counts / len(edf) * 100).round(1)
})
size_summary
```

```python
ax = size_counts.plot(kind="bar", figsize=(8, 5), color="violet")

ax.set_title("Product IDs by Size Variant")
ax.set_xlabel("Size variant")
ax.set_ylabel("Number of product IDs")
ax.bar_label(
    ax.containers[0],
    labels=[
        f"{count} ({count / len(edf) * 100:.1f}%)"
        for count in size_counts.values
    ], fontsize=10, 
    padding=3
)

plt.tight_layout()
plt.show()
```

<div style="background-color: purple;color:white; padding: 10px;">
<h3>I. Check for redundant marketplace price columns</h3>

Two columns are redundant if they have the same prices for every product and the same missing-value pattern. We check the Amazon MRP columns before deciding whether to keep both.
</div>

```python
amazon_price_pair = edf[["amazon_mrp", "amazon_fba_mrp"]]
amazon_paired_rows = amazon_price_pair.dropna()
amazon_prices_match = amazon_paired_rows["amazon_mrp"].eq(
    amazon_paired_rows["amazon_fba_mrp"]
).all()
amazon_missing_patterns_match = edf["amazon_mrp"].isna().equals(
    edf["amazon_fba_mrp"].isna()
)

amazon_redundancy_check = pd.DataFrame({
    "paired_product_ids": [len(amazon_paired_rows)],
    "matching_prices": [int(amazon_paired_rows["amazon_mrp"].eq(amazon_paired_rows["amazon_fba_mrp"]).sum())],
    "same_missing_pattern": [amazon_missing_patterns_match],
    "exact_duplicate_columns": [amazon_prices_match and amazon_missing_patterns_match]
}, index=["amazon_mrp vs amazon_fba_mrp"])
```

```python
amazon_redundancy_check.T
```

<div style="background-color: PeachPuff;color:black; padding: 10px;">
The check shows that <code>amazon_mrp</code> and <code>amazon_fba_mrp</code> contain identical values and missingness in this file. We retain <code>amazon_mrp</code> as the representative column and drop the duplicate <code>amazon_fba_mrp</code>. This removes duplicated information (which would be perfectly collinear if both were used as predictors); it does not imply other prices should be dropped merely because they are correlated.</div>

```python
# Drop the exact duplicate after documenting the check
edf = edf.drop(columns="amazon_fba_mrp")
```

<div style="background-color: purple;color:white; padding: 10px;"><h3>II. Check marketplace price coverage</h3>
Compare missing-value coverage across the original-price fields and every MRP listing column before removing the duplicate Amazon FBA column. The single <code style="color:aqua">tp</code> field is not included because the related schema describes TP 1 and TP 2, which are not present in this CSV.
</div>

```python
# check availability of prices in different marketplaces
mrp_cols = [
    "ajio_mrp", "amazon_mrp", "flipkart_mrp", "limeroad_mrp",
    "myntra_mrp", "paytm_mrp", "snapdeal_mrp"
]
mrp_coverage = pd.DataFrame({
    "product_ids_with_price": edf[mrp_cols].notna().sum(),
    "missing_prices": edf[mrp_cols].isna().sum(),
    "coverage_pct": edf[mrp_cols].notna().mean().mul(100).round(1)
})
mrp_coverage.index = (   # change the name of market price to make it pretty
    mrp_coverage.index
    .str.replace("_mrp", "", regex=False)
    .str.replace("_", " ", regex=False)
    .str.title()
)
rows_with_any_mrp = edf[mrp_cols].notna().any(axis=1).sum()
rows_with_no_mrp = edf[mrp_cols].isna().all(axis=1).sum()
print(f"Rows with at least one MRP value: {rows_with_any_mrp}")
print(f"Rows with no MRP values: {rows_with_no_mrp}")
mrp_coverage
```

<div style="background-color: purple;color:white; padding: 10px;"><h3>III. Compare marketplaces for the same product</h3>
For each product, the table below records how many marketplace prices are available and the range from the lowest to highest listed price. Products with fewer than two available prices are not eligible for this comparison.
</div>

```python
available_price_count = edf[mrp_cols].notna().sum(axis=1)
lowest_mrp_price = edf[mrp_cols].min(axis=1)
highest_mrp_price = edf[mrp_cols].max(axis=1)
mrp_price_range = highest_mrp_price - lowest_mrp_price
eligible_for_mrp_comparison = available_price_count.ge(2)

mrp_comparison = pd.DataFrame({
    "available_mrp_prices": available_price_count,
    "lowest_mrp_price": lowest_mrp_price,
    "highest_mrp_price": highest_mrp_price,
    "mrp_price_range": mrp_price_range
}, index=edf.index)

comparison_eligible = eligible_for_mrp_comparison
identical_prices = comparison_eligible & mrp_price_range.eq(0)
different_prices = comparison_eligible & mrp_price_range.gt(0)

mrp_consistency = pd.Series({                                                   # instead of printing several lines
    "Product IDs with at least two MRP prices": int(comparison_eligible.sum()),
    "Same price across available marketplaces": int(identical_prices.sum()),
    "At least one MRP price differs": int(different_prices.sum()),
    "Median price range among eligible products": mrp_price_range[comparison_eligible].median(),
    "Maximum price range": mrp_price_range[comparison_eligible].max()
}, name="result")
mrp_consistency
```

<div style="background-color: peachpuff;color:black; padding: 10px;"><h3><b>Top marketplace price spread</b></h3>
Products with the highest difference in price across different marketplaces.
</div>

```python
# Inspect the products with the widest spread between marketplace prices
top_marketplace_spreads = (
    edf.loc[different_prices, ["product_id", "style_id"]]
    .assign(
        lowest_mrp_price=lowest_mrp_price[different_prices],
        highest_mrp_price=highest_mrp_price[different_prices],
        mrp_price_range=mrp_price_range[different_prices]
    )
    .sort_values("mrp_price_range", ascending=False)
    .head(12)
)
top_marketplace_spreads
```

```python
# Distribution of non-zero price spreads across marketplaces for each product
nonzero_spreads = mrp_price_range[different_prices]

fig, ax = plt.subplots(figsize=(9, 5))
ax.hist(nonzero_spreads, bins=20, color="yellow", edgecolor="brown")
ax.set_title("Marketplace Price Spread for Products with Different Listings")
ax.set_xlabel("Highest minus lowest marketplace price")
ax.set_ylabel("Number of product IDs")
plt.tight_layout()
plt.show()
```

<div style="background-color:purple ;color:white; padding: 10px;">
    <h3>IV. Compare each marketplace price with the discounted price <code>final_mrp_old</code></h3>
We use <code>final_mrp_old</code> as a reference price and compare only product IDs that have both the reference and marketplace price. 'Lower' and 'higher' describe listed-price differences; they do not establish the price a customer paid or a realized discount.
</div>

```python
reference_price_col = "final_mrp_old"
comparison_rows = []

for marketplace in mrp_cols:
    paired_prices = edf[[reference_price_col, marketplace]].dropna()   # exclude if any of them is missing
    reference_prices = paired_prices[reference_price_col]
    marketplace_prices = paired_prices[marketplace]
    price_difference = marketplace_prices - reference_prices

    comparison_rows.append({
        "marketplace": marketplace.removesuffix("_mrp").replace("_", " ").title(),
        "num_products": len(paired_prices),
        "lower": int(price_difference.lt(0).sum()),          # marketplace price < final_mrp_old
        "same": int(price_difference.eq(0).sum()),           # marketplace price = final_mrp_old
        "higher": int(price_difference.gt(0).sum()),         # marketplace price > final_mrp_old
        "mean_difference": price_difference.mean().round(2), # Average (marketplace price - final_mrp_old); 
                                                             # positive = marketplace is higher, negative = marketplace is lower
    })

reference_comparison = pd.DataFrame(comparison_rows).set_index("marketplace")
reference_comparison.sort_values("mean_difference")
```

<div style="background-color: PeachPuff;color:black; padding: 10px;">
For most products, marketplace prices matched the reference discounted price <code>final_mrp_old</code>. When differences occurred, Amazon, Limeroad, Snapdeal, Ajio, Flipkart, and Paytm more often listed prices above the reference price than below it, while Myntra more frequently listed prices below the reference price.
</div>

<div style="background-color: purple;color:white; padding: 10px;"><h3>V. Visualize the comparisons of marketplace price and discounted price</h3>
The first chart compares how often each marketplace price is lower than, equal to, or higher than <code>final_mrp_old</code>. The second shows the size of the within-product price spread for products whose marketplace prices differ. These views highlight both the common pattern and the exceptions.
</div>


```python
# Display average difference of market price and discount price
marketplace_price_ranking = reference_comparison.sort_values("mean_difference")
marketplace_price_ranking.round(2)
ax = marketplace_price_ranking["mean_difference"].plot(
    kind="barh", figsize=(9, 4), color="darkcyan"
)
ax.axvline(0, color="black", linewidth=1)
ax.set_title("Average Marketplace Price Difference vs Final MRP Old")
ax.set_xlabel("Mean difference after discount - negative is lower, positive is higher")
ax.set_ylabel("Marketplace")
plt.tight_layout()
plt.show()
```

```python
# Stacked comparison counts of each marketplace prices against the reference price
position_data = reference_comparison[["lower", "same", "higher"]]

ax = position_data.plot(
    kind="barh", stacked=True, figsize=(14, 6),
    color=["green", "lightgray", "red"]
)
ax.set_title("Marketplace Listed Prices vs Final MRP Old")
ax.set_xlabel("Number of product IDs with both prices")
ax.set_ylabel("Marketplace")
ax.legend(title="Marketplace price", loc="center")
plt.tight_layout()
plt.show()
```

<div style="background-color: purple;color:white; padding: 10px;">
<h3>VI. Original price, final price after discount, and the discount amount</h3>
The source defines <code>mrp_old</code> as the original price and <code>final_mrp_old</code> as the price after discounts. Calculate the reduction as original minus final, and its percentage relative to the original. A negative reduction means the recorded final price is higher than the original, so it is not counted as a discount. We compare only products with both prices.
</div>


```python
discount_analysis = edf.dropna(subset=["mrp_old", "final_mrp_old"]).copy()
discount_analysis["discount_amount"] = (discount_analysis["mrp_old"] - discount_analysis["final_mrp_old"])
discount_analysis["discount_pct"] = (
    discount_analysis["discount_amount"]
    .div(discount_analysis["mrp_old"].where(discount_analysis["mrp_old"].ne(0)))
    .mul(100).round(1)
)

discount_summary = pd.Series({
    "Product IDs compared": len(discount_analysis),
    "Final price lower than original": int(discount_analysis["discount_amount"].gt(0).sum()),
    "Final price equal to original": int(discount_analysis["discount_amount"].eq(0).sum()),
    "Final price higher than original": int(discount_analysis["discount_amount"].lt(0).sum())
}, name="count")
discount_summary
```

<div style="background-color: PeachPuff;color:black; padding: 10px;">
<h3><b> Top 10 discounted products:</b></h3>
Note: The largest calculated reduction is an outlier worth checking against the source before presenting it as a genuine discount.
</div>


```python
discount_columns = ["product_id", "category", "catalog", "mrp_old","final_mrp_old", "discount_amount", "discount_pct"]
# Review the ten largest positive percentage reductions
top_discounts = (
    discount_analysis.loc[discount_analysis["discount_amount"].gt(0)]
    .nlargest(10, "discount_pct")[discount_columns]
)
top_discounts
```

<div style="background-color: purple;color:white; padding: 10px;">
<h3>VII. Listed prices by category and catalog</h3>
To compare product groups without mixing platform coverage into one average, reshape marketplace prices to one row per product-marketplace pair. The tables report median listed prices and the number of product IDs priced; the heatmaps visualize the median for each marketplace within each category or catalog. Missing category/catalog labels are grouped as <code>Unclassified</code>.
</div>


```python
from IPython.display import display

mrp_labels = {                                                        # Create readable marketplace names from column names such as
    column: column.removesuffix("_mrp").replace("_", " ").title()     # "amazon_mrp" -> "Amazon" and "flipkart_mrp" -> "Flipkart".
    for column in mrp_cols
}

marketplace_price_long = (                                          # Reshape the data from wide format (one column per marketplace)
    edf.assign(                                                     # to long format (one row per product-marketplace price).
        category=edf["category"].fillna("Unclassified"),            # This makes grouping and comparison across marketplaces easier.
        catalog=edf["catalog"].fillna("Unclassified")
    )
    .melt(                                                     # After melting, each marketplace price becomes a row rather than a separate column
        id_vars=["product_id", "category", "catalog"],
        value_vars=mrp_cols,
        var_name="marketplace",
        value_name="listed_price"
    )
    .dropna(subset=["listed_price"])
)
# Replace technical column names with cleaner marketplace labels.
marketplace_price_long["marketplace"] = marketplace_price_long["marketplace"].map(mrp_labels)

category_price_summary = marketplace_price_long.groupby(        # Aggregate marketplace prices within each category.
    ["category", "marketplace"]
).agg(
    median_listed_price=("listed_price", "median"),             # Median and mean summarize the central price level,
    mean_listed_price=("listed_price", "mean"),
    product_ids_priced=("product_id", "nunique")                # while nunique counts distinct products with a recorded price.
).round(1)

catalog_price_summary = marketplace_price_long.groupby(         # Aggregate marketplace prices within each catalog using
    ["catalog", "marketplace"]                                  # the same summary statistics.
).agg(
    median_listed_price=("listed_price", "median"),
    mean_listed_price=("listed_price", "mean"),
    product_ids_priced=("product_id", "nunique")
).round(1)

print("Category price summary:")
display(category_price_summary)
print("Catalog price summary:")
display(catalog_price_summary)
```

<div style="background-color: PeachPuff;color:black; padding: 10px;">
